```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

# Create pivot table
pivot_table = tips.pivot_table(
    index="day",
    columns="time",
    values="tip",
    aggfunc="mean"
)

print(pivot_table)

# Heatmap
sns.heatmap(
    pivot_table,
    annot=True,      # show values
    fmt=".2f",       # 2 decimal places
    cmap="YlGnBu"
)

plt.title("Average Tip by Day and Time")
plt.show()
```
